```python
from pathlib import Path

DATA_DIR = Path("../data/ABC_cleaned")

assert DATA_DIR.exists(), "Dataset directory not found"

# List all ABC files
abc_files = sorted(DATA_DIR.glob("*.abc"))

print(f"Number of ABC files: {len(abc_files)}")

for file in abc_files:
    print(file.name)
```

```python
jigs_path = DATA_DIR / "jigs.abc"

with open(jigs_path, "r", encoding="utf-8") as f:
    abc_text = f.read()

print(abc_text[:3000])
```

```python
import re

def extract_tunes(abc_text):
    """
    Split an ABC file containing multiple tunes
    into individual tune strings.
    """

    # Find each line beginning with "X:"
    parts = re.split(
        r"(?m)(?=^X\s*:)",
        abc_text
    )

    # Remove empty sections and whitespace
    tunes = [
        part.strip()
        for part in parts
        if re.match(r"^X\s*:", part.strip())
    ]

    return tunes


jigs_tunes = extract_tunes(abc_text)

print(f"Number of tunes in jigs.abc: {len(jigs_tunes)}")

# Display the first 2 tunes
print(f"FIRST TUNE:\n{jigs_tunes[0]}")
print(f"\nSECOND TUNE:\n{jigs_tunes[1]}")
```
