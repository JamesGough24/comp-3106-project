```python
from pathlib import Path

DATA_DIR = Path("../data/ABC_cleaned")

assert DATA_DIR.exists(), "Dataset directory not found"

# List all ABC files
abc_files = sorted(DATA_DIR.glob("*.abc"))

print(f"Number of ABC files: {len(abc_files)}")

for file in abc_files:
    print(file.name)
```

```python
jigs_path = DATA_DIR / "jigs.abc"

with open(jigs_path, "r", encoding="utf-8") as f:
    abc_text = f.read()

print(abc_text[:3000])
```

```python
import re

def extract_tunes(abc_text):
    """
    Split an ABC file containing multiple tunes
    into individual tune strings.
    """

    # Find each line beginning with "X:"
    parts = re.split(
        r"(?m)(?=^X\s*:)",
        abc_text
    )

    # Remove empty sections and whitespace
    tunes = [
        part.strip()
        for part in parts
        if re.match(r"^X\s*:", part.strip())
    ]

    return tunes


jigs_tunes = extract_tunes(abc_text)

print(f"Number of tunes in jigs.abc: {len(jigs_tunes)}")

# Display the first 2 tunes
print(f"FIRST TUNE:\n{jigs_tunes[0]}")
print(f"\nSECOND TUNE:\n{jigs_tunes[1]}")
```

```python
first_tune = jigs_tunes[0]

for line in first_tune.splitlines():
    if line.startswith("K:"):
        print("Key:", line)
    elif line.startswith("M:"):
        print("Meter:", line)
    elif line.startswith("L:"):
        print("Default note length:", line)
    elif line.startswith("X:"):
        print("Tune ID:", line)
    elif line.startswith("T:"):
        print("Title:", line)
```

```python
all_tunes = []

for file in abc_files:

    with open(file, "r", encoding="utf-8") as f:
        text = f.read()

    tunes = extract_tunes(text)

    print(f"{file.name}: {len(tunes)} tunes")

    for tune in tunes:
        all_tunes.append({
            "source_file": file.name,
            "abc": tune
        })

print("\nTotal tunes:", len(all_tunes))
```

```python
from collections import Counter
import re

field_counts = Counter()

for tune in all_tunes:
    for line in tune["abc"].splitlines():

        match = re.match(
            r"^\s*([A-Za-z]):",
            line
        )

        if match:
            field_counts[match.group(1)] += 1

print(field_counts.most_common())
```

```python
inline_fields = Counter()

for tune in all_tunes:
    text = tune["abc"]

    matches = re.findall(
        r"\[([A-Za-z]):",
        text
    )

    inline_fields.update(matches)

print(inline_fields.most_common())
```

```python
chord_count = 0

for tune in all_tunes:
    chord_count += len(
        re.findall(r'"[^"]*"', tune["abc"])
    )

print("Quoted annotations:", chord_count)
```

```python
# Inspect most common quoted annotations
from collections import Counter
import re

quoted_annotations = Counter()

for tune in all_tunes:
    annotations = re.findall(r'"([^"]*)"', tune["abc"])
    quoted_annotations.update(annotations)

print("Unique quoted annotations:", len(quoted_annotations))
print("\nMost common:")
for annotation, count in quoted_annotations.most_common(40):
    print(f"{annotation!r}: {count}")
```

```python
# Inspect musical symbols and notation patterns
special_patterns = {
    "Ties": r"-",
    "Slurs": r"[()]",
    "Grace notes": r"\{[^}]*\}",
    "Tuplets": r"\(\d+",
    "Broken rhythm": r"[<>]+",
    "Accidentals": r"[\^_=]+",
    "Rests": r"[zZxX]",
    "Bar lines": r"\|+|:\||\|:",
    "First/second endings": r"\[\d+",
    "Decorations": r"![^!]*!|\+[^+]*\+",
}

for name, pattern in special_patterns.items():
    count = 0

    for tune in all_tunes:
        # Remove quoted annotations so their contents don't interfere.
        text = re.sub(r'"[^"]*"', '', tune["abc"])
        count += len(re.findall(pattern, text))

    print(f"{name}: {count}")
```

```python
# Find examples of less familiar notation
patterns_to_inspect = {
    "Ties": r"-",
    "Slurs": r"[()]",
    "Grace notes": r"\{[^}]*\}",
    "Tuplets": r"\(\d+",
    "Broken rhythm": r"[<>]+",
    "Decorations": r"![^!]*!|\+[^+]*\+",
}

for name, pattern in patterns_to_inspect.items():
    examples = []

    for tune in all_tunes:
        text = re.sub(r'"[^"]*"', '', tune["abc"])

        if re.search(pattern, text):
            examples.append(text[:300])

        if len(examples) == 3:
            break

    print(f"\n--- {name} ---")
    for example in examples:
        print(example)
```

```python
from collections import Counter
import re

bracket_groups = Counter()

for tune in all_tunes:
    # Find bracketed groups, excluding repeat endings like [1 and [2.
    groups = re.findall(r"\[([^\]]*)\]", tune["abc"])

    for group in groups:
        if not re.fullmatch(r"\d+", group):
            bracket_groups[group] += 1

print("Unique bracketed groups:", len(bracket_groups))
print("\nMost common groups:")
for group, count in bracket_groups.most_common(40):
    print(f"{group!r}: {count}")
```

```python
tuplet_counts = Counter()

for tune in all_tunes:
    text = re.sub(r'"[^"]*"', '', tune["abc"])
    tuplet_counts.update(re.findall(r"\((\d+)", text))

print(tuplet_counts)
```

```python
import re
from collections import Counter

feature_counts = Counter()
feature_examples = {}

for tune in all_tunes:
    lines = tune["abc"].splitlines()

    # Find the first K: field, which marks the start of the ABC music body.
    key_index = next(
        (i for i, line in enumerate(lines)
         if re.match(r"^\s*K:", line)),
        None
    )

    if key_index is None:
        continue

    # Exclude subsequent standalone metadata fields and comments.
    body_lines = [
        line for line in lines[key_index + 1:]
        if not re.match(r"^\s*[A-Za-z]:", line)
        and not line.lstrip().startswith("%")
    ]

    body = "\n".join(body_lines)
    body = re.sub(r'"[^"]*"', '', body)

    features = {
        "simultaneous_notes": r"\[[^\]\d][^\]]*\]",
        "tuplets": r"\(\d+(?::\d+){0,2}",
        "ties": r"-",
        "slurs_or_tuplets": r"[()]",
        "decorations": r"![^!]*!",
    }

    for feature, pattern in features.items():
        matches = re.findall(pattern, body)

        if matches:
            feature_counts[feature] += 1
            feature_examples.setdefault(feature, matches[:5])

print("Number of tunes containing each feature:")
for feature, count in feature_counts.most_common():
    print(f"{feature}: {count} / {len(all_tunes)}")

print("\nExamples:")
for feature, examples in feature_examples.items():
    print(feature, examples)
```
